This is not good since regression (which will be used later) requires complete data. Since the null values will hinder the ability to use regression later, I use pandas .dropna to drop all NaN values. I figured that dropping only 207 values would not compromise the results. </p>

```python
#drop null values
data.dropna(inplace=True)
```

<p>This function validates all columns using regex to ensure that all columns are of type float64. If a column contains an invalid entry it is converted to the type of NaN. I use the pandas function .astype, .columns,.loc,.to_numeric  to accomplish this.  The column ocean_proximity is excluded because I will soon hot encode the values, thus they will soon change datatype.</p>

```python
def validateFloats(df, skip = None):
    # regex pattern for a valid float
    float_pattern = r'^-?\d+(\.\d+)?$'
    invalid_entries = {}
    for column in df.columns:
        if(column == skip):
            continue
        # match values in the column to the float regex
        valid_floats = df[column].astype(str).str.match(float_pattern, na=False)
        
        if valid_floats.all():
            print(f"Column '{column}' is valid and is float64.")
        else:
            print(f"Column '{column}' contains invalid entries.")
            invalid_entries[column] = df.loc[~valid_floats, column].tolist()
            print("Invalid values:", invalid_entries[column])
            # convert valid entries to float and set invalid ones to NaN
            df[column] = pd.to_numeric(df[column], errors='coerce')    
    return df, invalid_entries

```

```python
#validates data
validated_df, invalid_data = validateFloats(data, skip="ocean_proximity")
print("\nValidated DataFrame:")
print(validated_df)

print("\nInvalid Data Entries:")
print(invalid_data)
```

<p>Once that is complete, I use the .info() function again to verify that all variables have the same number of values.</p>

```python
#verifying changes, ensuring same amount of non null values
data.info()
```

<p>Additionally, I created the function check_out_of_bounds to identify any extreme values in the dataset. The columns minimum value, maximum value, mean value, and its standard deviation is outputted. Functions from numpy are used to calculate said values.</p>

```python
def check_out_of_bounds(df, numeric_columns=None):
    results = {}
    if numeric_columns:
        for col in numeric_columns:
            min_val = df[col].min()
            max_val = df[col].max()
            mean_val = df[col].mean()
            std_val = df[col].std()
            
            results[col] = {
                "min": min_val,
                "max": max_val,
                "mean": mean_val,
                "std_dev": std_val,
            }
            print(f"Column '{col}': Min={min_val}, Max={max_val}, Mean={mean_val}, Std Dev={std_val}")
    return results
```

<p>The results of the check out of bounds function are outputted. I did not note any extreme data that would compromise the results.</p>

```python
checkBoundsOf = ["housing_median_age","median_income","median_house_value", "households","total_rooms","total_bedrooms","population","households"]
check_out_of_bounds(data, numeric_columns=checkBoundsOf)
```

In order to successfully split my data, I use variables X and Y to split the features. X represents every feature except the target value (in this case, median_house_value), and y represents only the target value.<br>

<p>Histograms of the numerical features are plotted to analyse their distributions. It is observed that in every histogram there is at least one instance of skewed data. Matplotlib is used to print the histogram</p>

```python
# display histograms of numerical features to understand their distributions
train_data.hist(figsize=(15,8))
```

<p>In order to fix the skewed data, I apply a logarithmic transformation (log(x+1)). This works since the logarithmic function compresses large values and leaves smaller values relatively unchanged. This reduces the impact of extreme outliers in the dataset and modifies the dataframe to make the distribution closer to a normal distribution. I use numpy's .log function to accomplish this</p>

```python
#fix skewed data, apply a logarithmic transformation to reduce skewness
train_data['total_rooms'] = np.log(train_data['total_rooms']+1)
train_data['total_bedrooms'] = np.log(train_data['total_bedrooms']+1)
train_data['population'] = np.log(train_data['population']+1)
train_data['households'] = np.log(train_data['households']+1)
```

<p>As you can observe, the values distributions are much closer to a bell-shaped (normal) distribution than they were before.</p>

<p>The values of ocean_proximity are originally not numerical values. <h3>Feature engineering</h3>
<p>I made two features: bedroom ratio and household rooms. I did this to further explore how different rooms effect housing value. Bedroom ratio is calculated by dividing the median amount of bedrooms by the median amount of rooms. Similarly, household rooms are calculated by dividing the median total amount of rooms by the amount of households. I wanted to test the correlation between the bedroom ratio and household rooms to try to determine patterns in how space is allocated within homes.
There is a strong inverse relationship between bedroom_ratio and household_rooms. This means that as the number of rooms per house increases, the proportion of bedrooms in total decreases. In homes with a larger amount of rooms, likely a smaller fraction of those rooms are considered bedrooms. Larger homes may allocate more space to living areas, kitchens, or offices, thus reducing the ratio of bedrooms to total rooms.<br>
I additionally wanted to see how these new variables affected the median house value. The bedroom ratio has a negative correlation of -0.20. This means that as the bedroom ratio increases, the median house value decreases. In other words, a lower bedroom ratio indicates a higher-value home. This surprised me; I assumed that a higher bedroom ratio would indicate a larger home, which I assumed would cost more. Potential reasoning for the bedroom ratio increasing the median house value is that higher-value homes prioritise spaciousness and additional luxury over having a higher proportion of bedrooms.

However, a higher bedroom ratio has a positive correlation with the median amount of households. This could mean that a higher bedroom ratio could suggest that the housing is aimed at larger families or groups. The practical design of high bedroom ratios leads to more households in that area. This is because such homes are built to accommodate more households, thus accounting for the bedroom ratio being positively associated with the number of households.

</p>

```python
#Plot the updated correlation matrix including new engineered features
train_data['bedroom_ratio']= train_data['total_bedrooms']/train_data['total_rooms']
train_data['household_rooms'] = train_data['total_rooms']/train_data['households']
correlation_matrix = train_data.corr(numeric_only=True)
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix With New Engineered Features')
plt.show()
```

<p>This function sorts data into a given top and bottom percentile. This is done using pandas .quantile function </p>

```python
#sorts the top and bottom 10 percent of a given column
def sort_into_groups(df, column, group_size=0.10):
    top_cutoff = df[column].quantile(1 - group_size)
    bottom_cutoff = df[column].quantile(group_size)
    
    top_group = df[df[column] >= top_cutoff]
    bottom_group = df[df[column] <= bottom_cutoff]
    
    print(f"Top {group_size * 100}% cutoff for {column}: {top_cutoff}")
    print(f"Bottom {group_size * 100}% cutoff for {column}: {bottom_cutoff}")
    
    return top_group, bottom_group
```

This shows that houses within this category have a larger amount of rooms. There is a common room count (8.45) for houses in this category,followed by a decline as room count increases. Within the bottom 10% the histogram has a peak at 6.8 rooms which shows a high concentration of houses with that room count. The two distributions habe no overlap. The smaller size of rooms in the bottom 10% likely correlates with a lower house value, while the top 10% likely align with higher house value. Generally, the amount of rooms within a house seems to correlate with the house value.

```python
top_10_rooms, bottom_10_rooms = sort_into_groups(train_data, column = "total_rooms", group_size=0.10)
sns.histplot(top_10_rooms['total_rooms'], color='blue', label='Top 10%', kde=True, bins=30, alpha=0.6)
sns.histplot(bottom_10_rooms['total_rooms'], color='red', label='Bottom 10%', kde=True, bins=30, alpha=0.6)

plt.title("Distribution of Total Rooms (Top 10% vs. Bottom 10%)")
plt.xlabel("Total Rooms")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```
